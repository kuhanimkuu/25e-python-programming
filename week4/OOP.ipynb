# Object Oriented Programming

OOP is a programming approach that organizes code into 'objects' which are reusable and they contain both data (attributes) and behaviour(methods)

Instead of writing code like a list of instructions like we are used to, we instead create 'objects' which we can reuse in our program

## Key concepts in OOP

### 1. Class
A class is a blueprint, template or prototype from which objects are created
It defines the structure and behaviour that all objects of that type will have

### 2. Object
An object is an instance of a class. It is the concrete netity that exists in memory created based on the blueprint provided by the class.

### 3. Attributes
Attributes are data/characteristics that describe the objects. They represent the state of the object

### 4. Methods
Methods define what actions objects can do and what actions/operations can be done to them

## Principles of OOP

### Encapsulation
This is generally:
    Bundling data and behaviour together: putting an object'scharacteristics   and the actions it can perform into a single self-contained unit class
    Controlling access to data: Deciding what parts of that package are visible and directly changeable from the outside
