simple example of a class

```python
class Dog:
    species = 'german shephard'
    def bark(self):
        print('Woof woof')
```

'self' as an alias for the object- it represents specifically the current object. It allows each object to keep track of its own data. And it must be the first parameter of any method defined in a class, including the init method

```python
dog_1 = Dog()
```

```python
dog_1.bark()
print(dog_1.species)
```

#### Constructor in Phton (__init__ method)
Often you want each object to have its own unique set of initial attributes(xristics) when it is created
__init__ method is a special method that gives each object that will be created from the same calss, its own data

State is the value of an attribute
species is attribute, poodle is state

```python
class dog:
    legs = 4
    def __init__(self,species, name,age):
        self.species = species
        self.name = name
        self.age = age
    def bark(self):
        print("Woof Woof!")
```

```python
dog_1=dog("German Shephard", "Bosco",3)
dog_2=dog('Poodle','Karen',2)
```

```python
print(dog_1.species)
print(dog_2.species)
```

## Class variables vs Instance variables
class variable- attribute defined within the class but outside any methods eg legs above
Instance variable attribute defined within the class and inside one of the method(in the instance)
eg species in the above example

```python
class book:
    def __init__(self,title,author,book_no):
        self.title = title
        self.author = author
        self.book_no = book_no
        self.is_borrowed = False
    def borrow(self):
        self.is_borrowed=True
    def return_book(self):
        self.is_borrowed = False
    def __str__(self):
        status='Borrowed' if self.is_borrowed else 'available'
        return f'Title: {self.title}, Author: {self.author}, book_no:{self.book_no}, Status:{status}'
    

```

```python
book_1= book('the river between','Ngugi wan Thiongo',1)
```

```python
book_1.borrow()
```

```python
print(book_1)
```

```python
book_1.return_book()
print(book_1)
```

```python
class member:
    def __init__(self, name,member_id):
        self.name = name
        self.member_id = member_id
        self.borrow_book = []
    def borrow_books(self, book):
        if book.is_borrowed:
            print('Book is unavailable')
        else:
            book.borrow()
            self.borrow_book.append(book)
    def return_book(self,book):
        if book in self.borrow_book:
            self.borrow_book.remove(book)
            book.return_book()
    def __str__(self):
        books = ','.join([book.title for book in self.borrow_book]) or 'Member has no borrowed books'
        return f'Member: {self.name}, Member_id: {self.member_id}, Borrowed books: {books}'
    

  
```

```python
book_3 = book('Blossoms of the savannah','Henry Ole Kulet',1)
book_4 = book('Wihtout Fail','Lee Child',1)
member_1 = member('David',23)
member_1.borrow_books(book_3)
member_1.borrow_books(book_4)
```
