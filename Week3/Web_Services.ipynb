# XML

extensible markup language is a versatile platform independent language used for describing, storing and transporting data. It is widely used for data exchange between different systems and platforms due to its simplicity and flexibility.

## Key concepts
Elements
    The basic building blocks of an XML document.
    Defined by a start tag and an end tag.

Attributes:
    Provide additional information about elements.
    Defined within the start tag.

XML Declaration:
    The optional first line in an XML document.
    Specifies the XML version and character encoding.

Well-formed XML:
    XML documents must follow a set of rules to be considered well-formed:
        There must be a single root element.
        Tags must be properly nested.
        Tags must be properly closed.
        Attribute values must be enclosed in quotes.

Comments:
    Comments can be added within XML documents.
    They are enclosed within <!-- and -->

    Format Data as XML:
        Write a function to convert the weather data to XML format.

    Define XML Schema:
        Write an XML schema for the weather data.


```python
import random
import json
import xml.etree.ElementTree as ET
```

```python
city = str(input("Enter city name: "))
conditions = ['sunny','cloudy','rainy','calm']
weather = {'city':{}}
def create_weather():
    temperature = f'{random.randint(0,40)} degrees'
    humidity = f'{random.randint(0,100)} %'
    condition = conditions[random.randint(0,3)]
    weather['city']['name'] = city
    weather['city']['temperature'] = temperature
    weather['city']['humidity'] = humidity
    weather['city']['condition'] = condition
    #print(weather)

def weather_data_json ():
    weather_data = json.dumps(weather)
    print(weather_data)
def weather_data_xml():
    weather_data = ET.Element('weather')
    city_element = ET.SubElement(weather_data, "city")
    city_element.text = weather['city']['name']
    temperature_element = ET.SubElement(weather_data, "temperature")
    temperature_element.text = weather['city']['temperature']
    humidity_element = ET.SubElement(weather_data, "humidity")
    humidity_element.text = weather['city']['humidity']
    condition_element = ET.SubElement(weather_data, "condition")
    condition_element.text = weather['city']['condition']
    return ET.tostring(weather_data, encoding="utf-8").decode("utf-8")
create_weather()
print("Json data: ")
weather_data_json()

xml_data = weather_data_xml()

print("XML Data:")
print(xml_data)
```
