```python
from bs4 import BeautifulSoup
import requests
import csv

```

Webscrapping prices of airfriers from Jumia website

```python
url = 'https://www.jumia.co.ke/catalog/?q=airfryers'
headers = {
    'User-Agent':'Chrome/5.0'
}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.content,'html.parser')
products = soup.find_all('article', class_ = 'prd')

with open('air_fryers.csv', mode='w', newline='', encoding='utf-8' ) as file:
    writer = csv.writer(file)
    writer.writerow(['product_name', 'New_Price', 'Old_Price'])
    writer.writerows(products)

for product in products:
    name = product.find('h3',class_='name')
    new_price = product.find('div' ,class_ = 'prc')
    old_price = product.find('div', class_='old')

    if name and new_price:
        print(f'Product: {name.text.strip()}')
        print(f"New price: {new_price.text.strip()}")
        if old_price:
            print(f'Old price: {old_price.text.strip()}')
        else:
            print("Old price: Not listed")
        print('-'*40)
    
```

### Scrapping from multiple pages

Here we create another loop that will loop through the pages then everything else continues

```python


for page in range(1,7):
    url = 'https://www.jumia.co.ke/catalog/?q=airfryers&page={page}'
    headers = {
    'User-Agent':'Chrome/5.0'
    }
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.content,'html.parser')
    products = soup.find_all('article', class_ = 'prd')
    for product in products:
        name = product.find('h3',class_='name')
        new_price = product.find('div' ,class_ = 'prc')
        old_price = product.find('div', class_='old')

        if name and new_price:
            print(f'Product: {name.text.strip()}')
            print(f"New price: {new_price.text.strip()}")
            if old_price:
                print(f'Old price: {old_price.text.strip()}')
            else:
                print("Old price: Not listed")
            print('-'*40)
```
