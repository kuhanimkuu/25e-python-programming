```python
import requests
import pandas as pd
import json
```

```python
base_url = 'https://rickandmortyapi.com/api/'
end_point = "character"
response = requests.get(base_url + end_point)
response.raise_for_status()# this is kinda try except
data = json.loads(response.text)#loads the text as a json object
```

```python

print(json.dumps(data, indent=4))
location_name = data['results'][0]['location']['name']
character_name = data['results'][0]['name']
print(character_name)
print(location_name)
```

### Implementating Pagination (Getting all the characters)

```python
total_pages = data['info']['pages']

print(total_pages)
```

Getting name of all the characters
we loop through the pages using a for loop nad in each loop we get character name

```python
all_character_info = []
all_characters = []
for page in range(1, total_pages+1):
    base_url = base_url
    end_point = end_point
    url =f'{base_url}{end_point}?page={page}'
    response = requests.get(url)
    response.raise_for_status()
    data = json.loads(response.text)
    character_name = data['results'][0]['name']
    all_characters.append(character_name)
    print(character_name)
```

Getting a list of the locations


```python
character_location = {}
for page in range(1, total_pages+1):
    base_url = base_url
    end_point = end_point
    url = f'{base_url}{end_point}?page={page}'
    response = requests.get(url)
    response.raise_for_status()
    data = json.loads(response.text)
    character_name = data['results'][0]['name']
    location_name = data['results'][0]['location']['name']
    character_location.update({f'{character_name}':f'{location_name}'})

for k,v in character_location.items():
    print(f'Character name: {k}')
    print(f'Character_location: {v}')
    print('_'*50)
    print(" ")
    
```
